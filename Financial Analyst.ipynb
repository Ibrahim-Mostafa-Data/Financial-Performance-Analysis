# Data Preparation

```python
# Rename Columns :-
df_2014.rename(columns={'2015 PRICE VAR [%]':'PRICE VAR [%]'},inplace=True)
df_2015.rename(columns={'2016 PRICE VAR [%]':'PRICE VAR [%]'},inplace=True)
df_2016.rename(columns={'2017 PRICE VAR [%]':'PRICE VAR [%]'},inplace=True)
df_2017.rename(columns={'2018 PRICE VAR [%]':'PRICE VAR [%]'},inplace=True)
df_2018.rename(columns={'2019 PRICE VAR [%]':'PRICE VAR [%]'},inplace=True)
```

```python
# Add Year Column :-
df_2014['Year']=2014
df_2015['Year']=2015
df_2016['Year']=2016
df_2017['Year']=2017
df_2018['Year']=2018
```

```python
# Concatenate :-
financial_data = pd.concat([df_2014,df_2015,df_2016,df_2017,df_2018],ignore_index=True)
```
